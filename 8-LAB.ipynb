## LAB 8-1


* `datetime.now()`로 현재 날짜/시간 객체를 생성해 `now`에 저장
* `now.year`, `now.month`, `now.day`로 오늘 날짜를 출력
* `now.hour < 12` 조건으로 오전/오후를 판별해 12시간 형식으로 시간 출력

```python
from datetime import datetime
now = datetime.now()
print(f'오늘의 날짜 : {now.year}년 {now.month}월 {now.day}일')

if now.hour < 12:
    a = '오전'
    hour = now.hour
else:
    a = '오후'
    hour = now.hour - 12
print(f'현재시간 : {a} {hour}시 {now.minute}분 {now.second}초')
```

## LAB 8-2

* `datetime(년, 월, 일)`로 목표 날짜 객체를 생성
* `목표날짜 - today`로 두 날짜를 빼면 `timedelta` 객체 반환
* `gap.days`로 남은 일수, `gap.seconds // 3600`으로 남은 시간 출력

```python
from datetime import datetime

today = datetime.now()
print(f'오늘은 {today.year}년 {today.month}월 {today.day}일입니다.')

xmas = datetime(2026, 12, 25)
gap = xmas - today
print(f'2027년 크리스마스 까지는 {gap.days}일 {gap.seconds // 3600}시간 남았습니다.')

```

```python
from datetime import datetime

today = datetime.now()
print(f'오늘은 {today.year}년 {today.month}월 {today.day}일입니다.')

new_year = datetime(2027, 1, 1)
gap = new_year - today
print(f'2036년 새해 까지는 {gap.days}일 {gap.seconds // 3600}시간 남았습니다.')
```

## LAB 8-3

**문제 1.** `timedelta(days=1000)`을 사용해 오늘 날짜로부터 1,000일 후의 날짜를 계산하여 출력

**문제 2.** 사귄 날짜를 입력받아 `timedelta(days=100)`을 더해 100일 기념일을 계산하여 출력

```python
from datetime import datetime, timedelta
today = datetime.now()
future = today + timedelta(days=1000)
print(f'오늘 날짜 : {today.year}년 {today.month}월 {today.day}일')
print(f'1000일 후 : {future.year}년 {future.month}월 {future.day}일')
```

```python
from datetime import datetime, timedelta

date_couple = input('처음으로 사귄 연도와 월, 일을 입력하시오 : ')
year, month, day = map(int,date_couple.split())

first_day=  datetime(year, month, day)
hundred = first_day + timedelta(days=100)
print(f'100일 기념일은 : {hundred.year}년 {hundred.month}월 {hundred.day}일입니다.')
```

## LAB 8-6

**문제 1.** 코드 8-8에 `for _ in range(5)` 반복문을 추가해 로또 번호 5세트를 생성하여 출력

**문제 2.** `random.sample()`로 당첨 번호 6개와 보너스 번호 1개를 각각 추출하여 출력

```python
import random as rd

print('이번 주의 추천 로또 번호')
for i in range(5):                    # 5번 반복 (수정된 부분)
    lotto_list = list(range(1, 46))   # 1부터 45까지 생성
    rd.shuffle(lotto_list)            # 임의의 순서로 섞기
    lotto_list = lotto_list[:6]       # 앞 부분 6개만 선택
    lotto_list.sort()                 # 선택된 번호를 정렬
    print(lotto_list)
```

```python
import random

lotto = random.sample(range(1, 46), 6)
lotto.sort()
bonus = random.sample(range(1, 46), 1)

print('이번 주의 당첨 로또 번호 + 보너스')
print(f'{lotto} + {bonus}')
```
